```python
import matplotlib.pyplot as plt

plt.scatter(
    df['Total_Purchases'],
    df['Age']
)

plt.xlabel('Total_Purchases')
plt.ylabel('Age')
plt.title('Total Purchases vs. Age')
plt.show()
```
